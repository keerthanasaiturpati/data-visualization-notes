# Seaborn

A Python data visualization library built on top of matplotlib.

Matplotlib is for low-level control, where I customize every detail. Seaborn is for high-level work, where I want a good-looking plot quickly without much setup.

### Benefits
- Many built-in datasets are available (`tips`, `flights`, `penguins`)
- Many high-level features are handled automatically (colors, legends, averaging), unlike matplotlib
- Box plots, heatmaps, etc. can be made in a few lines

### Adding more info to a scatter plot

Same two columns (x and y), but we can make the dots carry extra info:

- `hue` → changes the **color** based on a column (e.g. smoker yes/no)
- `size` → changes the **dot size** based on a column (e.g. party size)
- `style` → changes the **dot shape** based on a column

In matplotlib I had to make a colors list by hand. Here I just give the column name and seaborn builds the colors and the legend itself.

````python
sns.relplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    kind = "line"
)

 # so this fid tells us that, scatter plot will be meaningful
````

### Heatmap needs a table shape

A heatmap draws a grid of colored squares, so it needs data in rows × columns form.
`flights` is a long list (one row per month per year), so I reshape it with `pivot`:

```python
flights.pivot(index="year", columns="month", values="passengers")
```

- `index` → what goes down the rows
- `columns` → what goes across
- `values` → the number that decides the color

Swapping `index` and `columns` just flips the plot.
- `cmap` → color scheme
- `annot=True` → writes the number inside each square
- `fmt="d"` → shows them as whole numbers

````python
flights.sample(15)
````

### Using seaborn with fig, ax

`fig, ax = plt.subplots()` makes the frame (fig) and the graph paper (ax).
Passing `ax=ax` tells seaborn which graph paper to draw on.

- seaborn draws the plot
- `ax.set_title()`, `ax.set_xlabel()`, `ax.set_ylabel()` label it
- `fig.tight_layout()` fixes the spacing (it belongs to `fig`, not `ax`)

For 2 plots side by side: `plt.subplots(1, 2)` gives an array `ax`, so use `ax[0]` and `ax[1]`.
With a single `plt.subplots()`, `ax` is just one object, so `ax[0]` gives "not subscriptable".

````python
fig, ax = plt.subplots(1, 2)   # 1 row, 2 columns

sns.lineplot(data=tips,
             x="day",
             y="total_bill",
             ax=ax[0])

sns.scatterplot(data=tips,
                x="total_bill",
                y="tip",
                ax=ax[1])

ax[0].set_title("Left plot")
ax[1].set_title("Right plot")

fig.tight_layout()
````

### Lineplot averages automatically

`tips` has many bills per day, but a line needs one point per day.
So seaborn groups the rows (by `day`, and by `sex` because of `hue`) and plots the **mean** `total_bill` of each group.

The shaded band around each line is a confidence interval (how sure we are about that average). `errorbar=None` removes it.

Old tutorials use `ci=None` for this. It's deprecated now, use `errorbar=None`.

To check the averaging myself:
```python
tips.groupby(["day", "sex"])["total_bill"].mean()
```
